Arithmatic Operations


```python
x = 10
y = 10
print ('x + y =', x + y) # Addition
print ('x - y =', x - y) # Subtraction
print ('x * y =', x * y) # Multiplication
print ('x / y =', x / y) # Division
print ('x // y =', x // y) # Floor Division
print ('x ** y =', x ** y) # Exponent (Power Operations)
print ('x % y =', x % y) # Modulus Operations
```

Order of Precedence  of arithmatic operators

```python
10 + 20 - 10 + 5 * 4 / 2
```

```python
10 + 20 - 10 + 5 ** 4 / 2 # Power Operator has higher priority
```

```python
10 + 20 - 10 + 5 % 4 / 2 # Modulus Operator has higher priority
```

Power > Modulus > Division > Multiplication > Subtraction > Addition

```python
x = 10
y = 15
print ('x > y is', x > y) # Greater than
print ('x < y is', x < y) # Less than
print ('x == y is', x == y) # Equal to
print ('x != y is', x != y) # Not equal to
print ('x >= y is', x >= y) # Greater than or equal to
print ('x <= y is', x <= y)
print ('x is y is', x is y) # Identity Operator
print ('x is not y is', x is not y) # Identity Operator
```
